```python
# A2PS-OUS reproduction -- paste each STEP block into its own Colab cell (or the whole file into one cell)


# ============================================================================
# # A²PS-OUS — full reproduction
# **Altitude Aware Adaptive Power Splitting with Opportunistic User Scheduling for Energy Harvesting UAV Relaying over Mixed α–κ–μ and Rayleigh Fading Channels**
#
# Run top to bottom (`Runtime → Run all`). Each step prints the computed values next to the paper's, and every figure is saved as a 300-dpi PNG in `figures/`.
#
# | Step | Produces |
# |---|---|
# | 1–2 | Imports, Table 1 parameters |
# | 3–4 | Geometry & channels, Eqs. (1)–(7) |
# | 5 | Saturating harvester and optimal split, Eqs. (8)–(13) |
# | 6 | Outage (14), Gauss–Laguerre (15), throughput (17), ABER (18), baselines (§3.10) |
# | 7 | Monte Carlo simulator + Algorithm 1 check |
# | 8 | Algorithm 2 – outage-optimal altitude (bounded Brent) |
# | 9–16 | Fig. 3, Fig. 4, Tables 2–3, Fig. 5, Fig. 6, Fig. 7 + Table 4, Fig. 8, Fig. 9 + Table 5 |
#
# Runtime on a standard Colab CPU: about 3–6 minutes.
# ============================================================================

# ============================================================================
# ## Step 1 — Imports and plot style
# ============================================================================

import numpy as np, os, time
import matplotlib.pyplot as plt
from scipy import special as sp, stats
from scipy.integrate import quad
from scipy.optimize import minimize_scalar, brentq
from scipy.special import comb

os.makedirs('figures', exist_ok=True)
plt.rcParams.update({
    'font.family': 'serif', 'font.serif': ['DejaVu Serif'], 'mathtext.fontset': 'dejavuserif',
    'font.size': 10, 'axes.grid': True, 'grid.alpha': 0.35, 'legend.fontsize': 8,
    'figure.dpi': 110, 'savefig.dpi': 300, 'savefig.bbox': 'tight'})

# ============================================================================
# ## Step 2 — Simulation parameters (Table 1)
# Distances are normalised to $d_0 = 100$ m; ρ is the transmit SNR referred to $d_0$. $P_{max} = 0.03\,P_S$, so $P_{max}/P_S = 0.03$.
# ============================================================================

d0 = 100.0                      # reference distance (m)
rS0, ru0, H0 = 300.0, 50.0, 100.0   # source horizontal distance, user ring radius, default altitude
ae, be = 4.88, 0.43             # suburban A2G constants [7]
bN, bL, bRD = 3.5, 2.0, 2.7     # first-hop NLoS/LoS exponents [8], second-hop exponent
k0, k90 = 10**(5/10), 10**(15/10)   # Rician factor 5 dB (grazing) and 15 dB (vertical)
ALPHA, MU = 2.5, 1.5            # default alpha-kappa-mu parameters
N0 = 4                          # number of users
eta, Pmax_rel = 0.75, 0.03      # EH efficiency, P_max / P_S
R_t = 1.0
g_th = 2**(2*R_t) - 1           # = 3 (4.77 dB)
LAMS, AT = (0.75, 0.5), 0.3     # fixed PSR ratios, TSR fraction
g_th_tsr = 2**(2*R_t/(1 - AT)) - 1
H_MIN, H_MAX = 20.0, 300.0
NMC = 300_000                   # Monte Carlo realisations per point
print(f'gamma_th = {g_th} ({10*np.log10(g_th):.2f} dB), TSR threshold = {g_th_tsr:.3f}')

# ============================================================================
# ## Step 3 — Geometry and elevation-dependent parameters (Eqs. 1–3)
# $P_{LoS}$ uses θ in degrees, κ(θ) uses θ in radians.
# ============================================================================

def geom(H, rS=rS0, ru=ru0):
    th = np.arctan2(H, rS)
    d_SR = np.hypot(rS, H)/d0
    d_n = np.hypot(ru, H)/d0
    PL = 1/(1 + ae*np.exp(-be*(np.degrees(th) - ae)))
    beta = bN + (bL - bN)*PL
    kap = k0*(k90/k0)**(2*th/np.pi)
    return th, d_SR, d_n, beta, kap

th, d_SR, d_n, beta, kap = geom(H0)
print(f'H = 100 m: theta = {np.degrees(th):.1f} deg, beta = {beta:.2f}, kappa = {kap:.2f}   (paper 18.4, 2.02, 5.07)')

# ============================================================================
# ## Step 4 — First hop α–κ–μ (Eqs. 4–6) and second hop with opportunistic scheduling (Eq. 7)
# $1-Q_\mu(a,b)$ is evaluated as the non-central χ² CDF at $b^2$ with $2\mu$ degrees of freedom and non-centrality $a^2$. $Y=\max_n|g_n|^2$ has $F_Y(y)=(1-e^{-y})^N$.
# ============================================================================

def xhat(kap, al=ALPHA, mu=MU):
    num = sp.gamma(mu + 2/al)*np.exp(-kap*mu)*sp.hyp1f1(mu + 2/al, mu, kap*mu)
    return sp.gamma(mu)*(mu*(1 + kap))**(2/al)/num

def F_X(x, kap, al=ALPHA, mu=MU):                      # Eq. (5)
    x = np.maximum(x, 0.0)
    b2 = 2*mu*(1 + kap)*(x/xhat(kap, al, mu))**(al/2)
    return stats.ncx2.cdf(b2, 2*mu, 2*kap*mu)

def sample_X(rng, n, kap, al=ALPHA, mu=MU):
    V = rng.noncentral_chisquare(2*mu, 2*kap*mu, n)
    return xhat(kap, al, mu)*(V/(2*mu*(1 + kap)))**(2/al)

F_Y = lambda y, N: (1 - np.exp(-y))**N                 # Eq. (7)
f_Y = lambda y, N: N*np.exp(-y)*(1 - np.exp(-y))**(N - 1)

rng = np.random.default_rng(1); Xs = sample_X(rng, 10**6, kap)
print('E[X] (MC) = %.4f;  CDF check at x=0.5: analysis %.4f vs MC %.4f' % (Xs.mean(), F_X(0.5, kap), (Xs <= 0.5).mean()))

# ============================================================================
# ## Step 5 — Power-splitting relay with saturating harvester and optimal split (Eqs. 8–13)
# $a=\rho/d_{SR}^{\beta}$, $b=\eta\rho/(d_{SR}^{\beta}d_n^{\beta_{RD}})$, $e=\rho P_{max}/(P_S d_n^{\beta_{RD}})$.
# $\gamma_1=(1-\lambda)aX$, $\gamma_2=\min\{\lambda bXY,\,eY\}$; optimum $\lambda^\star=\min\{a/(a+bY),\lambda_{sat}\}$, $\gamma^\star_{SD}=\min\{abXY/(a+bY),\,eY\}$.
# ============================================================================

def consts(rho_dB, H=H0, rS=rS0, ru=ru0):
    rho = 10**(rho_dB/10)
    th, d_SR, d_n, beta, kap = geom(H, rS, ru)
    a = rho/d_SR**beta
    b = eta*rho/(d_SR**beta*d_n**bRD)
    e = rho*Pmax_rel/d_n**bRD
    return a, b, e, kap

def lam_star(a, b, e, X, Y):                           # Eq. (12); lambda_sat = e/(bX)
    return np.minimum(a/(a + b*Y), e/(b*X))

def gamma_sd(lam, a, b, e, X, Y):                      # Eq. (11), DF min
    return np.minimum((1 - lam)*a*X, np.minimum(lam*b*X*Y, e*Y))

# Proposition 1 check: brute-force search over lambda vs closed form
a, b, e, kap = consts(25); rng = np.random.default_rng(3)
X, Y = sample_X(rng, 5, kap), rng.exponential(1, (5, N0)).max(1)
lg = np.linspace(0, 1, 200001)
for x, y in zip(X, Y):
    brute = gamma_sd(lg, a, b, e, x, y).max()
    closed = min(a*b*x*y/(a + b*y), e*y)
    print(f'X={x:.3f} Y={y:.3f}: lambda*={lam_star(a,b,e,x,y):.4f}  brute max={brute:.5f}  Eq.(13)={closed:.5f}')

# ============================================================================
# ## Step 6 — Analysis: outage (14), Gauss–Laguerre (15), throughput (17), ABER (18) and baselines (§3.10)
# The outage is evaluated as $F_Y(\gamma_{th}/e)+\int_{\gamma_{th}/e}^{\infty}F_X(x_{th}(y))f_Y(y)\,dy$, which is algebraically identical to (14) and avoids cancellation at very small outage.
# Baselines: fixed PSR (constant λ), TSR ($\alpha_t=0.3$, relay power $\min\{2\eta\alpha_t P_{in}/(1-\alpha_t),P_{max}\}$, threshold $2^{2R/(1-\alpha_t)}-1$), adaptive PS with random user ($N=1$), adaptive PS with linear EH ($P_{max}\to\infty$).
# ============================================================================

SCHEMES = ['prop', 'lin', 'fix0.75', 'fix0.5', 'tsr', 'rand']
LABEL = {'prop': 'Proposed A²PS-OUS', 'lin': 'Adaptive PS, linear EH (bound)', 'fix0.75': r'Fixed PSR, $\lambda$=0.75',
         'fix0.5': r'Fixed PSR, $\lambda$=0.5', 'tsr': r'TSR, $\alpha_t$=0.3', 'rand': 'Adaptive PS, random user'}
STY = {'prop': dict(color='tab:red', ls='-', marker='o'), 'lin': dict(color='k', ls=':', marker=None),
       'fix0.75': dict(color='tab:blue', ls='--', marker='s'), 'fix0.5': dict(color='tab:cyan', ls='--', marker='D'),
       'tsr': dict(color='tab:purple', ls='-.', marker='^'), 'rand': dict(color='tab:gray', ls='--', marker='v')}
thr = lambda s: g_th_tsr if s == 'tsr' else g_th
PLAIN = {'prop': 'Proposed A2PS-OUS', 'lin': 'Adaptive PS, linear EH (bound)', 'fix0.75': 'Fixed PSR, lambda=0.75',
         'fix0.5': 'Fixed PSR, lambda=0.5', 'tsr': 'TSR, alpha_t=0.3', 'rand': 'Adaptive PS, random user'}
prelog = lambda s: (1 - AT)/2 if s == 'tsr' else 0.5

def pout(rho_dB, N=N0, H=H0, rS=rS0, scheme='prop', lam=None, al=ALPHA, mu=MU, g=None):
    a, b, e, kap = consts(rho_dB, H, rS)
    if scheme.startswith('fix'):
        lam = float(scheme[3:]) if lam is None else lam; scheme = 'fix'
    g = thr(scheme) if g is None else g
    if scheme == 'rand': N = 1
    if scheme in ('prop', 'rand'):
        y0, xth = g/e, lambda y: g*(a + b*y)/(a*b*y)
    elif scheme == 'lin':
        y0, xth = 0.0, lambda y: g*(a + b*y)/(a*b*y)
    elif scheme == 'fix':
        y0, xth = g/e, lambda y: np.maximum(g/((1 - lam)*a), g/(lam*b*y))
    elif scheme == 'tsr':
        c = 2*AT/(1 - AT)
        y0, xth = g/e, lambda y: np.maximum(g/a, g/(c*b*y))
    f = lambda y: F_X(xth(y), kap, al, mu)*f_Y(y, N)
    I = sum(quad(f, lo, hi, limit=200, epsabs=1e-17, epsrel=1e-9)[0]
            for lo, hi in [(y0, y0 + 2), (y0 + 2, y0 + 10), (y0 + 10, y0 + 60)])
    return F_Y(y0, N) + I

def pout_gl(rho_dB, N=N0, H=H0, rS=rS0, K=30):         # Eq. (15), Gauss-Laguerre
    a, b, e, kap = consts(rho_dB, H, rS)
    t, w = np.polynomial.laguerre.laggauss(K)
    y0 = g_th/e; s = 0.0
    for i in range(1, N + 1):
        y = y0 + t/i
        s += comb(N, i)*(-1)**(i + 1)*np.exp(-i*y0)*np.sum(w*(1 - F_X(g_th*(a + b*y)/(a*b*y), kap)))
    return 1 - s

def erg(rho_dB, scheme='prop', **kw):                  # Eq. (17)
    f = lambda g: (1 - pout(rho_dB, scheme=scheme, g=g, **kw))/(1 + g)
    return 2*prelog(scheme)/(2*np.log(2))*quad(f, 0, np.inf, limit=200)[0]

def aber(rho_dB, scheme='prop', A=1.0, B=0.5, **kw):   # Eq. (18), BPSK
    f = lambda g: np.exp(-A*g)*g**(B - 1)*pout(rho_dB, scheme=scheme, g=g, **kw)
    return A**B/(2*sp.gamma(B))*quad(f, 0, np.inf, limit=200)[0]

for N, ref in [(1, 0.356), (2, 0.129), (4, 0.0199), (8, 0.0025)]:
    print(f'N={N}: Pout(25 dB) = {pout(25, N):.4f}  (paper {ref})')
print(f'Gauss-Laguerre (15), K=30, N=4, 25 dB: {pout_gl(25):.4f}')
print(f'Ergodic (17) at 25 dB: {erg(25):.3f} bps/Hz (paper 1.782);  ABER (18) at 25 dB: {aber(25):.3e} (paper 6.24e-4)')

# ============================================================================
# ## Step 7 — Monte Carlo simulator (Algorithm 1 for every block)
# Each block draws $X$ (α–κ–μ) and $|g_n|^2$ (exponential), applies the per-block split $\lambda_k^\star$ of Algorithm 1 and best-user selection, and records $\gamma_{SD}$ for every scheme on the **same** channel draws.
# ============================================================================

def mc(rho_dB, N=N0, H=H0, rS=rS0, n=NMC, seed=0, al=ALPHA, mu=MU):
    rng = np.random.default_rng(seed)
    a, b, e, kap = consts(rho_dB, H, rS)
    X = sample_X(rng, n, kap, al, mu)
    G = rng.exponential(1, (n, N))
    Y, Yr = G.max(1), G[:, 0]                       # best user / a random user
    out = {'prop': gamma_sd(lam_star(a, b, e, X, Y), a, b, e, X, Y),     # Algorithm 1
           'lin':  a*b*X*Y/(a + b*Y),
           'rand': gamma_sd(lam_star(a, b, e, X, Yr), a, b, e, X, Yr)}
    for l in LAMS:
        out[f'fix{l}'] = gamma_sd(l, a, b, e, X, Y)
    out['tsr'] = np.minimum(a*X, np.minimum(2*AT/(1 - AT)*b*X*Y, e*Y))
    return out

def metrics(g, s):
    po = np.mean(g < thr(s))
    return dict(pout=po, dl=(1 - po)*R_t, erg=prelog(s)*np.mean(np.log2(1 + g)),
                ber=np.mean(0.5*sp.erfc(np.sqrt(g))))

t = time.time(); r = mc(25)
print({s: round(metrics(r[s], s)['pout'], 4) for s in SCHEMES}, f'({time.time()-t:.1f} s)')

# Run the whole SNR grid once; reused by Figs. 3, 4, 6, 8 and Tables 2-3
rho_mc = np.arange(15, 45.01, 2.5); rho_line = np.linspace(15, 45, 61)
t = time.time()
MC = {r: {s: metrics(g, s) for s, g in mc(r).items()} for r in rho_mc}
print(f'MC grid done ({time.time()-t:.1f} s)')

# ============================================================================
# ## Step 8 — Algorithm 2: outage-optimal altitude (Eq. 19)
# Bounded Brent search (golden section + parabolic interpolation) on $J(H)=\log_{10}P_{out}(H)$ over $[20, 300]$ m.
# ============================================================================

def opt_altitude(rho_dB, rS=rS0, N=N0, tol=1e-3):
    res = minimize_scalar(lambda H: np.log10(pout(rho_dB, N, H, rS)), bounds=(H_MIN, H_MAX),
                          method='bounded', options=dict(xatol=tol))
    return res.x, 10**res.fun

ALT = {}
for rS, rho in [(300, 30), (300, 35), (400, 35), (400, 40)]:
    ALT[(rS, rho)] = opt_altitude(rho, rS)
    print(f'rS={rS} m, rho={rho} dB -> H* = {ALT[(rS,rho)][0]:.1f} m, Pout = {ALT[(rS,rho)][1]:.3e}')

# ============================================================================
# ## Step 9 — Fig. 3: outage of A²PS-OUS for N = 1, 2, 4, 8 (analysis + MC)
# ============================================================================

fig, ax = plt.subplots(figsize=(5.4, 4.2))
cN = {1: 'tab:blue', 2: 'tab:green', 4: 'tab:red', 8: 'tab:purple'}
for N in (1, 2, 4, 8):
    ax.semilogy(rho_line, [pout(r, N) for r in rho_line], color=cN[N], label=f'N = {N} (analysis)')
    v = np.array([np.mean(mc(r, N=N)['prop'] < g_th) for r in rho_mc])
    ax.semilogy(rho_mc[v > 0], v[v > 0], 'o', mfc='none', color=cN[N])
ax.plot([], [], 'ko', mfc='none', label='Monte Carlo')
ax.set(ylim=(1e-7, 1.5), xlim=(14, 46), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel=r'Outage probability $P_{out}$')
ax.legend(loc='lower left'); fig.savefig('figures/fig3_outage_users.png'); plt.show()
for N, ref in [(1, 0.99), (2, 1.98), (4, 2.01), (8, 1.91)]:
    sl = -(np.log10(pout(45, N)) - np.log10(pout(37.5, N)))/0.75
    print(f'N={N}: slope 37.5-45 dB = {sl:.2f} (paper {ref}), G_d = {min(ALPHA*MU/2, N)}')

# ============================================================================
# ## Step 10 — Fig. 4: outage of all schemes (N = 4, H = 100 m)
# ============================================================================

fig, ax = plt.subplots(figsize=(5.4, 4.2))
for s in SCHEMES:
    v = np.array([MC[r][s]['pout'] for r in rho_mc]); m = v > 0
    st = STY[s]; ax.semilogy(rho_mc[m], v[m], ls=st['ls'], marker=st['marker'], ms=5, color=st['color'], label=LABEL[s])
ax.set(ylim=(1e-6, 1.5), xlim=(14, 46), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel=r'Outage probability $P_{out}$')
ax.legend(loc='lower left'); fig.savefig('figures/fig4_baselines.png'); plt.show()

# ============================================================================
# ## Step 11 — Table 2 (performance at 25 and 30 dB) and Table 3 (SNR for $P_{out}=10^{-2}$)
# Table 2 is Monte Carlo; the analytical outage is also printed. For Table 3 the exact analytical root and the Monte Carlo grid estimate are both shown — the paper's values are MC read off the 2.5 dB grid, which varies by about ±0.05 dB with the random seed.
# ============================================================================

REF2 = {'prop': (0.0199, 4.10e-4, 1.782, 2.565, 6.24e-4), 'fix0.75': (0.0440, 1.00e-3, 1.466, 2.223, 1.45e-3),
        'fix0.5': (0.0288, 1.00e-3, 1.720, 2.496, 9.44e-4), 'tsr': (0.1297, 4.61e-3, 1.297, 1.847, 5.35e-4),
        'rand': (0.3565, 0.1290, 1.215, 1.893, 3.29e-2), 'lin': (0.0079, 2.43e-4, 1.944, 2.735, 2.75e-4)}
print('Table 2 (MC; analysis outage in brackets; paper values on the second line)')
print(f"{'Scheme':32s}{'Pout 25':>20s}{'Pout 30':>22s}{'erg 25':>9s}{'erg 30':>9s}{'ABER 25':>11s}")
for s in ['prop', 'fix0.75', 'fix0.5', 'tsr', 'rand', 'lin']:
    a25, a30 = pout(25, scheme=s), pout(30, scheme=s)
    print(f"{PLAIN[s]:32s}{MC[25.0][s]['pout']:>9.4f} [{a25:.4f}]{MC[30.0][s]['pout']:>11.3e} [{a30:.2e}]"
          f"{MC[25.0][s]['erg']:>9.3f}{MC[30.0][s]['erg']:>9.3f}{MC[25.0][s]['ber']:>11.2e}")
    p = REF2[s]; print(f"{'   paper':32s}{p[0]:>9.4f}{'':9s}{p[1]:>11.3e}{'':11s}{p[2]:>9.3f}{p[3]:>9.3f}{p[4]:>11.2e}")

def snr_at(target, vals, grid=rho_mc):
    y = np.log10(np.maximum(vals, 1e-12)); return np.interp(np.log10(target), y[::-1], grid[::-1])
REF3 = {'prop': 25.93, 'fix0.75': 26.90, 'fix0.5': 26.67, 'tsr': 28.91, 'rand': 41.32, 'lin': 24.67}
print('\nTable 3: SNR for Pout = 1e-2')
root = {s: brentq(lambda r: np.log10(pout(r, scheme=s)) + 2, 10, 60) for s in REF3}
mcg = {s: snr_at(1e-2, np.array([MC[r][s]['pout'] for r in rho_mc])) for s in REF3}
print(f"{'Scheme':32s}{'exact root':>11s}{'MC grid':>9s}{'paper':>8s}{'gain (dB)':>11s}")
for s in REF3:
    print(f"{PLAIN[s]:32s}{root[s]:>11.2f}{mcg[s]:>9.2f}{REF3[s]:>8.2f}{root[s]-root['prop']:>11.2f}")

# ============================================================================
# ## Step 12 — Fig. 5: outage vs fixed power-splitting ratio λ (25, 30, 35 dB)
# ============================================================================

lams = np.round(np.arange(0.05, 0.951, 0.05), 2)
fig, ax = plt.subplots(figsize=(5.4, 4.2))
for rho, c in [(25, 'tab:blue'), (30, 'tab:green'), (35, 'tab:red')]:
    rng = np.random.default_rng(0); a, b, e, kap = consts(rho)
    X = sample_X(rng, NMC, kap); Y = rng.exponential(1, (NMC, N0)).max(1)
    v = np.array([np.mean(gamma_sd(l, a, b, e, X, Y) < g_th) for l in lams])
    ax.semilogy(lams, v, 's-', ms=4, color=c, label=fr'Fixed $\lambda$, $\rho$={rho} dB')
    ad = pout(rho); ax.axhline(ad, ls='--', color=c, lw=1.2)
    an = np.array([pout(rho, scheme='fix', lam=l) for l in lams]); i = np.argmin(an)
    print(f'rho={rho} dB: best fixed lambda (analysis) = {lams[i]} -> Pout {an[i]:.3g};  lambda=0.6 -> {an[11]:.3g} (MC {v[11]:.3g});  '
          f'adaptive = {ad:.3g}  ({100*(1-ad/an[i]):.0f}% lower)')
ax.plot([], [], 'k--', label=r'Proposed adaptive $\lambda_k^\star$')
ax.set(ylim=(1e-5, 1.5), xlabel=r'Power-splitting ratio $\lambda$', ylabel=r'Outage probability $P_{out}$')
ax.legend(loc='upper center'); fig.savefig('figures/fig5_lambda.png'); plt.show()
print('Paper: best 0.6 (0.0260) at 25 dB, 0.6 (7.25e-4) at 30 dB, 0.5 (~1.5e-5) at 35 dB; reductions 24% and 36%')
print('Note: at 35 dB only ~3-10 outage events occur per MC point, so the red MC curve is jagged (as in the paper).')

# ============================================================================
# ## Step 13 — Fig. 6: (a) delay-limited and (b) ergodic throughput
# ============================================================================

fig, (a1, a2) = plt.subplots(1, 2, figsize=(10, 3.9))
for s in SCHEMES:
    st = STY[s]; kw = dict(ls=st['ls'], marker=st['marker'], ms=4, color=st['color'], label=LABEL[s])
    a1.plot(rho_mc, [MC[r][s]['dl'] for r in rho_mc], **kw)
    a2.plot(rho_mc, [MC[r][s]['erg'] for r in rho_mc], **kw)
a1.set(xlabel=r'Transmit SNR $\rho$ (dB)', ylabel='Delay-limited throughput (bps/Hz)', title=r'(a) Delay-limited, $R$ = 1 bps/Hz')
a2.set(xlabel=r'Transmit SNR $\rho$ (dB)', ylabel='Ergodic throughput (bps/Hz)', title='(b) Ergodic'); a2.legend(loc='upper left')
fig.tight_layout(); fig.savefig('figures/fig6_throughput.png'); plt.show()
print(f"DL at 20 dB: prop {MC[20.0]['prop']['dl']:.2f}, fix0.75 {MC[20.0]['fix0.75']['dl']:.2f}, TSR {MC[20.0]['tsr']['dl']:.2f} "
      f"(paper 0.62, 0.18, 0.18); prop at 25 dB {MC[25.0]['prop']['dl']:.2f} (paper 0.98)")
e25 = {s: MC[25.0][s]['erg'] for s in SCHEMES}
print('Ergodic gain of A²PS-OUS at 25 dB: ' + ', '.join(f"{s} {100*(e25['prop']/e25[s]-1):+.1f}%" for s in ['fix0.75', 'fix0.5', 'tsr', 'rand'])
      + '  (paper +21.6%, +3.6%, +37.5%, +46.7%)')

# ============================================================================
# ## Step 14 — Fig. 7 and Table 4: effect of UAV altitude
# ============================================================================

Hs = np.linspace(20, 300, 141)
fig, ax = plt.subplots(figsize=(5.4, 4.2))
cases = [(300, 30, 'tab:red'), (300, 35, 'tab:orange'), (400, 35, 'tab:blue'), (400, 40, 'tab:green')]
for rS, rho, c in cases:
    ax.semilogy(Hs, [pout(rho, H=H, rS=rS) for H in Hs], color=c, label=fr'$r_S$={rS} m, $\rho$={rho} dB')
    Hst, pst = ALT[(rS, rho)]; ax.plot(Hst, pst, '*', ms=13, color=c, mec='k')
ax.plot([], [], 'k*', ms=11, label=r'$H^\star$ (Alg. 2)')
ax.set(ylim=(1e-6, 1.5), xlabel=r'UAV altitude $H$ (m)', ylabel=r'Outage probability $P_{out}$')
ax.legend(loc='lower right'); fig.savefig('figures/fig7_altitude.png'); plt.show()

print(f"{'rS':>4}{'rho':>5}{'H*':>8}{'theta*':>8}{'beta*':>7}{'kappa*':>8}{'Pout(H*)':>12}{'Pout(100)':>12}{'reduction':>11}")
for rS, rho, _ in cases:
    Hst, pst = ALT[(rS, rho)]; th_, _, _, b_, k_ = geom(Hst, rS); p100 = pout(rho, H=100, rS=rS)
    print(f"{rS:>4}{rho:>5}{Hst:>8.1f}{np.degrees(th_):>8.1f}{b_:>7.2f}{k_:>8.2f}{pst:>12.3e}{p100:>12.3e}{100*(1-pst/p100):>10.1f}%")
print('Paper: 73.4/13.7/1.93e-4/4.61e-4/58.1% | 83.5/15.6/1.23e-5/1.50e-5/18.2% | 98.0/13.8/1.108e-4/1.111e-4/0.3% | 106.4/14.9/8.13e-6/8.29e-6/2.0%')

# ============================================================================
# ## Step 15 — Fig. 8: ABER (BPSK), analysis (18) + Monte Carlo
# ============================================================================

fig, ax = plt.subplots(figsize=(5.4, 4.2))
ax.semilogy(rho_line, [aber(r) for r in rho_line], color='tab:red', label='Proposed (analysis)')
v = np.array([MC[r]['prop']['ber'] for r in rho_mc])
ax.semilogy(rho_mc, v, 'o', mfc='none', color='tab:red', label='Proposed A²PS-OUS (MC)')
for s in ['lin', 'fix0.75', 'fix0.5', 'tsr', 'rand']:
    st = STY[s]; v = np.array([MC[r][s]['ber'] for r in rho_mc]); m = v > 1e-9
    ax.semilogy(rho_mc[m], v[m], ls=st['ls'], marker=st['marker'], ms=5, color=st['color'], label=LABEL[s])
ax.set(ylim=(1e-7, 0.5), xlim=(14, 46), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel='ABER (BPSK)')
ax.legend(loc='lower left', fontsize=7); fig.savefig('figures/fig8_aber.png'); plt.show()
b = {s: MC[25.0][s]['ber'] for s in SCHEMES}
print(f"ABER at 25 dB: analysis {aber(25):.3e}, MC {b['prop']:.3e} (paper 6.24e-4); "
      f"{100*(1-b['prop']/b['fix0.75']):.0f}% below lambda=0.75 (paper 57%), {100*(1-b['prop']/b['fix0.5']):.0f}% below lambda=0.5 (paper 34%)")
print(f"Random user at 40 dB: {MC[40.0]['rand']['ber']:.2e} (paper: above 1e-3)")

# ============================================================================
# ## Step 16 — Fig. 9 and Table 5: effect of first-hop α–κ–μ parameters
# ============================================================================

fig, ax = plt.subplots(figsize=(5.4, 4.2))
for (al, mu), c in [((2, 1), 'tab:gray'), ((2.5, 1.5), 'tab:red'), ((3, 2), 'tab:blue'), ((4, 2), 'tab:green')]:
    ax.semilogy(rho_mc, [pout(r, al=al, mu=mu) for r in rho_mc], 'o-', ms=4, color=c, label=fr'$\alpha$={al:g}, $\mu$={mu:g}')
ax.set(ylim=(1e-9, 2), xlim=(14, 46), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel=r'Outage probability $P_{out}$')
ax.legend(loc='lower left'); fig.savefig('figures/fig9_akm.png'); plt.show()

REF5 = [(2.5, 1.5, 1, 0.99, 0.129), (2.5, 1.5, 2, 1.98, 0.0170), (2.5, 1.5, 8, 1.91, 8.3e-5), (2, 1, 4, 1.08, 5.8e-3),
        (2.5, 1.5, 4, 2.01, 4.6e-4), (3, 2, 4, 3.91, 2.6e-4), (4, 2, 4, 3.98, 2.6e-4)]
print(f"{'alpha':>6}{'mu':>5}{'N':>3}{'G_d':>7}{'slope':>8}{'paper':>7}{'Pout30':>11}{'paper':>10}")
for al, mu, N, sref, pref in REF5:
    sl = -(np.log10(pout(45, N, al=al, mu=mu)) - np.log10(pout(37.5, N, al=al, mu=mu)))/0.75
    print(f"{al:>6g}{mu:>5g}{N:>3}{min(al*mu/2, N):>7.3f}{sl:>8.2f}{sref:>7.2f}{pout(30, N, al=al, mu=mu):>11.2e}{pref:>10.2g}")

# ============================================================================
# ## Step 17 — Download all figures (Colab)
# ============================================================================

import shutil
shutil.make_archive('A2PS_OUS_figures', 'zip', 'figures')
try:
    from google.colab import files; files.download('A2PS_OUS_figures.zip')
except ImportError:
    print('Saved A2PS_OUS_figures.zip')
```
